```python
# Equação quadrática
import math
a = float(input("valor de a:"))
b = float(input("valor de b:"))
c = float(input("valor de c:"))
print((a,b,c))
d = (b**2)-4*a*c
print()
if a == 0 and b != 0:
    x = -c/b
    print("Equação de primeiro grau: x= %.2f"%(x))
elif a == 0 and b == 0:
    if c == 0:
        print("Infinitas soluções: qualquer valor de x.")
    else:
        print("Não existe solução.")
elif d > 0:
    x1 = (-b + math.sqrt(d))/(2*a)
    x2 = (-b - math.sqrt(d))/(2*a)
    print("Duas raízes reais distintas: x1 = %.2f x2 = %.2f"%(x1,x2))
elif d == 0:
    x1 = (-b + math.sqrt(d))/(2*a)
    x2 = (-b - math.sqrt(d))/(2*a)
    print("Uma raíz real dupla: x1 = x2 %.2f" %(x1))
else:
    import cmath
    x1 = (-b + cmath.sqrt(d))/(2*a)
    x2 = (-b - cmath.sqrt(d))/(2*a)
    print("Nenhuma raiz real: x1 = %s x2= %s"%(x1,x2)) # %.2f não aceita números complexos.

import matplotlib.pyplot as plt
import numpy as np
valores_x = np.linspace(-1, 4, 400)
valores_y = a*(valores_x)**2 + b*valores_x + c

fig = plt.figure()

plt.plot(valores_x, valores_y, color ="red")
plt.axhline(0, color="gray")
plt.axvline(0, color="gray")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gráfico de y = ax² + bx + c")
plt.grid(True)

plt.show()
```
